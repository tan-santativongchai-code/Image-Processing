```python
import glob
import os
import cv2
import numpy as np
```

```python
# 1. Define your input and output folders
input_folder = r'D:\MTEC\Image-Processing\Cell Classification and Counting\cell detection dataset\Unprocessed\100x'  # Folder containing your original images
output_folder = (
    r'D:\MTEC\Image-Processing\Cell Classification and Counting\cell detection dataset\Processed\100x'  # New folder where corrected images will be saved
)

# 2. Create the output folder if it doesn't exist
os.makedirs(output_folder, exist_ok=True)
```

```python
# 3. Find all common image formats in the input folder
# (Microscopy images are often saved as .tif, .png, or .jpg)
image_extensions = ['*.jpg', '*.jpeg', '*.png', '*.tif', '*.tiff', '*.bmp']
image_paths = []
for ext in image_extensions:
  image_paths.extend(glob.glob(os.path.join(input_folder, ext)))
```

```python
if not image_paths:
  print(f"No images found in '{input_folder}'. Please check your folder path.")
else:
  print(f'Found {len(image_paths)} images to process...')

  # 4. Loop through each image and apply processing
  for path in image_paths:
    filename = os.path.basename(path)

    # Load image in grayscale
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)

    if img is None:
      print(f'Skipping {filename}: Could not read image.')
      continue

    # Step A: Uneven illumination correction (bright center flattening)
    background = cv2.GaussianBlur(img, (151, 151), 0)
    corrected = cv2.divide(img, background, scale=255)

    # Step B: Enhance cell visibility using CLAHE
    clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8, 8))
    enhanced = clahe.apply(corrected)

    # Step C: Optional sharpening to pop cell boundaries
    kernel = np.array([[0, -1, 0], [-1, 5, -1], [0, -1, 0]])
    sharpened = cv2.filter2D(enhanced, -1, kernel)

    base_name, ext = os.path.splitext(filename)
    if ext.lower() in ['.tif', '.tiff']:
      ext = '.jpg'  # Automatically convert raw tiff files to roboflow-friendly jpg

    output_filename = base_name + ext
    output_path = os.path.join(output_folder, output_filename)

    cv2.imwrite(output_path, sharpened)
    print(f'Processed and saved: {filename}')

  print('Batch processing complete!')
```
